# Audit timestamped tool interval interface

This notebook loads `audit_v<source>` with `taxonomy_v<source>`, shows HTML image panels for each clip, and displays matching `audit_v<source>_camera_segments` sidecar context when available.

Compared with `audit_version_interface.ipynb`, actions are organized by actor/tool timing: camera actions keep per-action intervals plus a movement description, left-instrument actions are listed under **Left** with retraction action intervals, changed/start/end direction, and description, and right-instrument actions are listed under **Right** with tool, action, target, target context, interval, and active-tool description rows.

Existing right-instrument annotations with multiple `target_structure_*` values are split into separate right interval rows. For now, each split target repeats the same two target-context fields from the original annotation.

Set `SOURCE_VERSION` and optionally `TARGET_VERSION` below. If `TARGET_VERSION = None`, the UI automatically saves into the first missing `audit_vN` directory after the source version.

Use the `Jump to` box in the launched UI to load any 1-based video index. If that video already has a saved file in the target audit directory, the saved target annotations are loaded so they can be edited and saved again.


```python
import os
import sys

ROOT = os.getcwd()
if os.path.basename(ROOT) == 'notebooks':
    ROOT = os.path.dirname(ROOT)
SRC = os.path.join(ROOT, 'src')
if SRC not in sys.path:
    sys.path.insert(0, SRC)

from cvs_act.audit_timestamped_tool_interface import AuditVersionInterface

SOURCE_VERSION = 10
TARGET_VERSION = 11  # set an int like 11, or leave None to auto-pick the next free audit_vN
```
