# Experiments to run

## Stage A — Prediction performance (keep prompts constant; no rationale)

Use pred-only schema everywhere so prompt budget isn’t a confound.

A1. Single frame, no organs (vanilla_pred_only)
baseline.

A2. Single frame + organ overlay (vanilla_organs_pred_only)
isolates overlay effect.

A3. Sequence, no organs (sequential_pred_only)
isolates temporal context effect.

A4. Sequence + organs (sequential_organs_pred_only)
tests whether overlay helps beyond temporal.

That’s a clean 2×2:

input: single vs sequence

overlay: off vs on

✅ 4 conditions only.

If you also have decision history: don’t add it yet. Keep it off, or keep it fixed (e.g., always included but empty) so it’s not another dimension.

## Stage B — Explanation faithfulness (only on the best 1–2 models)

Now you test your explanation protocol without multiplying conditions.

Pick the best-performing input configuration from Stage A (often A4, but let results decide).

Then compare:

B1. pred-only (no rationale)
B2. obs-then-decide (rationale + decisions)
(optional) B3. decomposed-Q (sub-answers → final)

✅ Here you’re isolating protocol effect while holding vision input constant.

This gives you a story like:

“Given the same visual input, structured prompting yields more faithful, teachable explanations (and may slightly affect accuracy).”
